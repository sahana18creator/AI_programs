```python

def is_variable(term):
    return term[0].islower()


def parse_expression(expression):
    expression = expression.strip()
    name, arguments = expression.split("(", 1)
    arguments = arguments.rstrip(")")
    args = [arg.strip() for arg in arguments.split(",")]
    return name.strip(), args


def apply_substitution(term, substitutions):
    while term in substitutions:
        term = substitutions[term]
    return term


def unify_terms(a, b, substitutions):
    a = apply_substitution(a, substitutions)
    b = apply_substitution(b, substitutions)

    if a == b:
        return True

    if is_variable(a):
        substitutions[a] = b
        return True

    if is_variable(b):
        substitutions[b] = a
        return True

    return False


def unify(expr1, expr2):
    pred1, args1 = parse_expression(expr1)
    pred2, args2 = parse_expression(expr2)

    if pred1 != pred2:
        return None

    if len(args1) != len(args2):
        return None

    substitutions = {}

    for a, b in zip(args1, args2):
        if not unify_terms(a, b, substitutions):
            return None

    return substitutions


# Main program
expr1 = input("Enter first expression: ")
expr2 = input("Enter second expression: ")

result = unify(expr1, expr2)

if result is None:
    print("Unification Failed")
elif len(result) == 0:
    print("Expressions are already identical.")
    print("MGU: NIL")
else:
    print("Unification Successful!")
    print("Substitutions:")
    for variable, value in result.items():
        print(variable, "=", value)

    print("MGU:", result)
```
