```python

import math
import random

# User inputs
current_cost = int(input("Enter current cost: "))
next_cost = int(input("Enter next cost: "))
temperature = float(input("Enter temperature: "))

# Calculate energy difference
delta_E = current_cost - next_cost

print("\nEnergy Difference:", delta_E)

if delta_E > 0:
    print("Better solution: Accepted")

else:
    # Calculate acceptance probability
    probability = math.exp(delta_E / temperature)

    print("Acceptance Probability:", round(probability, 4))

    # Generate random number
    r = random.random()

    print("Random Number:", round(r, 4))

    if r < probability:
        print("Worse solution accepted")
    else:
        print("Worse solution rejected")
```
