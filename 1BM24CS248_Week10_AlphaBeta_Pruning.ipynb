```python

def alpha_beta(tree, alpha, beta, maximizing):

    if isinstance(tree, int):
        return tree

    if maximizing:
        value = float('-inf')

        for i, child in enumerate(tree):
            value = max(value, alpha_beta(child, alpha, beta, False))

            alpha = max(alpha, value)

            print("MAX: Alpha =", alpha, "Beta =", beta)

            if alpha >= beta:
                print("Pruned subtree:", tree[i + 1:])
                break

        return value

    else:
        value = float('inf')

        for i, child in enumerate(tree):
            value = min(value, alpha_beta(child, alpha, beta, True))

            beta = min(beta, value)

            print("MIN: Alpha =", alpha, "Beta =", beta)

            if alpha >= beta:
                print("Pruned subtree:", tree[i + 1:])
                break

        return value


def main():

    print("Enter 8 leaf node values:")

    values = list(map(int, input().split()))

    if len(values) != 8:
        print("Please enter exactly 8 values.")
        return

    # Construct the game tree
    tree = [
        [
            [values[0], values[1]],
            [values[2], values[3]]
        ],
        [
            [values[4], values[5]],
            [values[6], values[7]]
        ]
    ]

    print("\nGame Tree:", tree)

    result = alpha_beta(
        tree,
        float('-inf'),
        float('inf'),
        True
    )

    print("\nOptimal value of Root MAX node:", result)


if __name__ == "__main__":
    main()
```
