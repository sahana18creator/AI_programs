```python

# Initial Facts in Knowledge Base
facts = {
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Owns", "A", "T1"),
    ("Missile", "T1")
}

# Forward Chaining
changed = True

while changed:
    changed = False

    # Rule 1: Missile(x) -> Weapon(x)
    for fact in list(facts):
        if fact[0] == "Missile":
            new_fact = ("Weapon", fact[1])

            if new_fact not in facts:
                facts.add(new_fact)
                changed = True

    # Rule 2: Enemy(x, America) -> Hostile(x)
    for fact in list(facts):
        if fact[0] == "Enemy" and fact[2] == "America":
            new_fact = ("Hostile", fact[1])

            if new_fact not in facts:
                facts.add(new_fact)
                changed = True

    # Rule 3: Missile(x) AND Owns(A,x)
    # -> Sells(Robert,x,A)
    for m in list(facts):
        for o in list(facts):
            if (m[0] == "Missile" and
                o[0] == "Owns" and
                m[1] == o[2] and o[1] == "A"):

                new_fact = ("Sells", "Robert", m[1], "A")

                if new_fact not in facts:
                    facts.add(new_fact)
                    changed = True

    # Rule 4: Crime rule
    for a in list(facts):
        for w in list(facts):
            for s in list(facts):
                for h in list(facts):

                    if (a[0] == "American" and
                        w[0] == "Weapon" and
                        s[0] == "Sells" and
                        h[0] == "Hostile" and
                        a[1] == s[1] and
                        w[1] == s[2] and
                        s[3] == h[1]):

                        new_fact = ("Criminal", a[1])

                        if new_fact not in facts:
                            facts.add(new_fact)
                            changed = True

# Display Knowledge Base
print("Knowledge Base:")
for fact in sorted(facts):
    print(fact)

# Check the Query
query = ("Criminal", "Robert")

print("\nQuery: Criminal(Robert)")

if query in facts:
    print("Robert is Criminal")
    print("Query Proved: TRUE")
else:
    print("Query Not Proved: FALSE")
```
