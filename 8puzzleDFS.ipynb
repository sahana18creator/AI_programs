```python
import copy

GOAL_STATE = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]  # 0 represents the empty tile
]

def find_zero(board):
    """Find the (row, col) position of the empty space (0)."""
    for r in range(3):
        for c in range(3):
            if board[r][c] == 0:
                return r, c

def get_neighbors(state):
    """Generate all valid next states from the current state."""
    r, c = find_zero(state)
    moves = [(-1, 0, "UP"), (1, 0, "DOWN"), (0, -1, "LEFT"), (0, 1, "RIGHT")]
    neighbors = []

    for dr, dc, move_name in moves:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_board = copy.deepcopy(state)
            new_board[r][c], new_board[nr][nc] = new_board[nr][nc], new_board[r][c]
            neighbors.append((new_board, move_name))

    return neighbors

def dfs_8_puzzle(initial_state, max_depth=30):
    """
    Solves 8-puzzle using DFS.
    Stores (current_state, path_of_moves, path_of_boards) in the stack.
    """
    # Stack stores: (current_state, moves_list, board_history_list)
    stack = [(initial_state, [], [initial_state])]
    visited = set()

    while stack:
        current_state, moves, history = stack.pop()
        state_tuple = tuple(tuple(row) for row in current_state)

        if current_state == GOAL_STATE:
            return moves, history

        if state_tuple not in visited:
            visited.add(state_tuple)

            if len(moves) < max_depth:
                for neighbor, move in get_neighbors(current_state):
                    neighbor_tuple = tuple(tuple(row) for row in neighbor)
                    if neighbor_tuple not in visited:
                        stack.append((
                            neighbor,
                            moves + [move],
                            history + [neighbor]
                        ))

    return None, None

def print_board(board):
    """Prints a 3x3 board in a formatted grid."""
    print("+---+---+---+")
    for row in board:
        row_str = "| " + " | ".join(str(val) if val != 0 else " " for val in row) + " |"
        print(row_str)
        print("+---+---+---+")

def get_user_input():
    """Takes 9 space-separated numbers from the user and constructs a 3x3 board."""
    print("Enter 9 numbers (0-8) separated by spaces for the 3x3 board.")
    print("Use '0' to represent the empty tile (e.g., '1 2 3 4 0 6 7 5 8'):")

    while True:
        try:
            raw_input = input(">> ").strip().split()
            nums = [int(x) for x in raw_input]

            if len(nums) != 9:
                print("Error: Please enter exactly 9 numbers.\n")
                continue

            if sorted(nums) != list(range(9)):
                print("Error: Numbers must contain all digits from 0 to 8 without duplicates.\n")
                continue

            # Reshape 1D list of 9 elements into 3x3 2D list
            board = [nums[i:i + 3] for i in range(0, 9, 3)]
            return board

        except ValueError:
            print("Error: Invalid input! Please enter integers only.\n")

# --- Execution Flow ---
if __name__ == "__main__":
    initial_board = get_user_input()

    print("\nInitial State:")
    print_board(initial_board)

    print("Searching for solution using DFS...\n")
    moves, history = dfs_8_puzzle(initial_board, max_depth=20)

    if moves is not None:
        print(f"Goal state reached in {len(moves)} move(s)!\n")
        print("================ STEP-BY-STEP SOLUTION ================\n")

        # Display each step with move name and resulting board state
        for idx, (board, move) in enumerate(zip(history[1:], moves), start=1):
            print(f"Step {idx}: Move blank space {move}")
            print_board(board)
            print()

        print("Puzzle Solved!")
    else:
        print("No solution found within the maximum depth limit.")
```

```python
import copy

GOAL_STATE = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]


def find_zero(board):
    """Find the position of the blank space."""
    for r in range(3):
        for c in range(3):
            if board[r][c] == 0:
                return r, c


def get_neighbors(state):
    """Generate all possible next states."""

    r, c = find_zero(state)

    moves = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]

    neighbors = []

    for dr, dc, move_name in moves:

        nr = r + dr
        nc = c + dc

        if 0 <= nr < 3 and 0 <= nc < 3:

            new_board = copy.deepcopy(state)

            # Swap blank with the tile
            new_board[r][c], new_board[nr][nc] = \
                new_board[nr][nc], new_board[r][c]

            neighbors.append((new_board, move_name))

    return neighbors


def depth_limited_search(state, depth, limit, path, visited):
    """
    DFS with a depth limit.
    """

    # Check goal
    if state == GOAL_STATE:
        return path

    # Stop if depth limit is reached
    if depth == limit:
        return None

    state_tuple = tuple(tuple(row) for row in state)

    visited.add(state_tuple)

    # Explore neighbors
    for neighbor, move in get_neighbors(state):

        neighbor_tuple = tuple(tuple(row) for row in neighbor)

        if neighbor_tuple not in visited:

            result = depth_limited_search(
                neighbor,
                depth + 1,
                limit,
                path + [(move, neighbor)],
                visited
            )

            if result is not None:
                return result

    return None


def ids_8_puzzle(initial_state, max_depth=20):

    for limit in range(max_depth + 1):

        print("Searching with depth limit:", limit)

        visited = set()

        result = depth_limited_search(
            initial_state,
            0,
            limit,
            [],
            visited
        )

        if result is not None:
            return result, limit

    return None, None


def print_board(board):

    print("+---+---+---+")

    for row in board:

        row_str = "| " + " | ".join(
            str(value) if value != 0 else " "
            for value in row
        ) + " |"

        print(row_str)
        print("+---+---+---+")


def get_user_input():

    print("Enter 9 numbers from 0 to 8.")
    print("Use 0 for the empty tile.")
    print("Example: 1 2 3 4 0 6 7 5 8")

    while True:

        try:

            numbers = list(
                map(int, input(">> ").split())
            )

            if len(numbers) != 9:
                print("Please enter exactly 9 numbers.")
                continue

            if sorted(numbers) != list(range(9)):
                print("Numbers must contain 0 to 8 exactly once.")
                continue

            board = [
                numbers[i:i + 3]
                for i in range(0, 9, 3)
            ]

            return board

        except ValueError:

            print("Please enter integers only.")


# -------------------------------
# Main Program
# -------------------------------

if __name__ == "__main__":

    initial_board = get_user_input()

    print("\nInitial State:")
    print_board(initial_board)

    print("\nSearching using IDS...\n")

    solution, depth = ids_8_puzzle(
        initial_board,
        max_depth=20
    )

    if solution is not None:

        print("\nGoal state reached!")
        print("Solution depth:", depth)

        print("\n========== STEP-BY-STEP SOLUTION ==========\n")

        for step, (move, board) in enumerate(
            solution,
            start=1
        ):

            print(
                f"Step {step}: Move blank space {move}"
            )

            print_board(board)
            print()

        print("Puzzle Solved!")

    else:

        print("No solution found.")
```
