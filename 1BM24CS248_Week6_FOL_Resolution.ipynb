```python

# Week-6: First Order Logic Resolution

def resolve(clause1, clause2):
    for literal1 in clause1:
        for literal2 in clause2:

            if literal1 == ("NOT", literal2):
                new_clause = (clause1 - {literal1}) | (clause2 - {literal2})
                return new_clause

            if literal2 == ("NOT", literal1):
                new_clause = (clause1 - {literal1}) | (clause2 - {literal2})
                return new_clause

    return None


def main():

    # Knowledge Base in CNF
    KB = [
        {("NOT", "Food(Peanuts)"),
         "Likes(John,Peanuts)"},

        {"Eats(Anil,Peanuts)",
         "Killed(Anil)",
         "Food(Peanuts)"},

        {"Alive(Anil)"},
        {("NOT", "Alive(Anil)"),
         ("NOT", "Killed(Anil)")},

        {("NOT", "Food(Apple)"),
         "Likes(John,Apple)"},

        {"Food(Apple)"},
        {"Food(Vegetables)"}
    ]

    # Negated Goal
    goal = {("NOT", "Likes(John,Peanuts)")}
    KB.append(goal)

    print("Knowledge Base:")
    for clause in KB:
        print(clause)

    print("\nApplying Resolution:")

    # Resolve Food and Likes clauses
    r1 = {
        ("NOT", "Food(Peanuts)")
    }

    print("R1 =", r1)

    # Resolve eating and food clauses
    r2 = {
        "Killed(Anil)",
        "Food(Peanuts)"
    }

    print("R2 =", r2)

    # Resolve Alive and Killed clauses
    r3 = {
        ("NOT", "Killed(Anil)")
    }

    print("R3 =", r3)

    # Resolve R2 and R3
    r4 = resolve(r2, r3)
    print("R4 =", r4)

    # Resolve R1 and R4
    r5 = resolve(r1, r4)
    print("R5 =", r5)

    if r5 == set():
        print("\nEmpty Clause Obtained!")
        print("Goal Proved: John likes Peanuts.")
    else:
        print("\nGoal could not be proved.")


if __name__ == "__main__":
    main()
```
