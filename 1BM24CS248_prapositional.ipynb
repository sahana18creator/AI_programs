```python

from itertools import product


def implies(a, b):
    return (not a) or b


def check_kb(P, Q, R):
    s1 = implies(Q, P)        # Q -> P
    s2 = implies(P, not Q)    # P -> not Q
    s3 = Q or R               # Q or R

    kb = s1 and s2 and s3
    return s1, s2, s3, kb


def truth_table():
    print("Queries:")
    print("1. R")
    print("2. R -> P")
    print("3. Q -> R")

    choice = int(input("Enter your query choice (1-3): "))

    queries = {
        1: ("R", lambda P, Q, R: R),
        2: ("R -> P", lambda P, Q, R: implies(R, P)),
        3: ("Q -> R", lambda P, Q, R: implies(Q, R))
    }

    if choice not in queries:
        print("Invalid choice!")
        return

    name, query = queries[choice]

    print("\nP Q R | Q->P P->~Q Q|R | KB | Query")
    print("-" * 48)

    entailed = True

    for P, Q, R in product([False, True], repeat=3):
        s1, s2, s3, kb = check_kb(P, Q, R)
        q_value = query(P, Q, R)

        tf = lambda x: "T" if x else "F"

        print(
            f"{tf(P)} {tf(Q)} {tf(R)} |  "
            f"{tf(s1)}    {tf(s2)}    {tf(s3)}  | "
            f"{tf(kb)}  | {tf(q_value)}"
        )

        if kb and not q_value:
            entailed = False

    print("\nQuery:", name)

    if entailed:
        print("KB entails", name, ": True")
    else:
        print("KB does not entail", name, ": False")


truth_table()
```
