```python
from itertools import product

states = list(product(["A", "B"], ["D", "C"], ["D", "C"]))

print("All Possible States:")
for i, state in enumerate(states, 1):
    print(f"{i}. Vacuum: {state[0]}, Room A: {state[1]}, Room B: {state[2]}")

def goal_test(state):
    position, room_a, room_b = state
    return room_a == "C" and room_b == "C"

def vacuum_agent(state):
    position, room_a, room_b = state

    if goal_test(state):
        return "STOP"

    if position == "A":
        if room_a == "D":
            return "CLEAN"
        else:
            return "MOVE_RIGHT"

    if position == "B":
        if room_b == "D":
            return "CLEAN"
        else:
            return "MOVE_LEFT"

def apply_action(state, action):
    position, room_a, room_b = state

    if action == "CLEAN":
        if position == "A":
            room_a = "C"
        elif position == "B":
            room_b = "C"

    elif action == "MOVE_RIGHT":
        position = "B"

    elif action == "MOVE_LEFT":
        position = "A"

    return (position, room_a, room_b)

def run_vacuum(initial_state):
    state = initial_state

    print("\nInitial State:")
    print(state)

    step = 0

    while not goal_test(state):
        step += 1
        action = vacuum_agent(state)

        print(f"\nStep {step}")
        print("Current State:", state)
        print("Action:", action)

        state = apply_action(state, action)

        print("New State:", state)

    print("\nGoal State Reached!")
    print("Both rooms are clean.")
    print("Final State:", state)

initial_state = ("A", "D", "D")

run_vacuum(initial_state)
```
