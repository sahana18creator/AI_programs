```python

def cost(state):
    conflicts = 0

    for i in range(4):
        for j in range(i + 1, 4):
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def hill_climbing():
    state = list(map(int, input(
        "Enter row positions of 4 queens (0-3): "
    ).split()))

    if len(state) != 4 or sorted(state) != [0, 1, 2, 3]:
        print("Invalid input! Enter a permutation of 0, 1, 2, 3.")
        return

    print("\nInitial State:", state)
    print("Initial Cost:", cost(state))

    while cost(state) != 0:
        neighbors = []

        for i in range(4):
            for j in range(i + 1, 4):
                next_state = state.copy()
                next_state[i], next_state[j] = (
                    next_state[j], next_state[i]
                )
                neighbors.append(
                    (cost(next_state), i, j, next_state)
                )

        print("\nNeighbor States:")
        for value, i, j, next_state in neighbors:
            print(f"Swap ({i},{j}): {next_state}, Cost = {value}")

        best = min(neighbors, key=lambda x: (x[0], x[1], x[2]))

        if best[0] >= cost(state):
            print("\nNo better neighbor found.")
            print("Final State:", state)
            print("Final Cost:", cost(state))
            return

        state = best[3]
        print("\nSelected State:", state)
        print("Selected Cost:", cost(state))

    print("\nGoal State Reached!")
    print("Final State:", state)
    print("Final Cost:", cost(state))


hill_climbing()
```
